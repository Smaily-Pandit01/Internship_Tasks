```python
# Q1. Find the Shortest Word in a Sentence

sentence = "Python makes coding easy"

# 1. Observe : Input - Python makes coding easy, Output - easy
# 2. Understanding : To find out a shortest word from the above input.
# 3. Breakdown : a) Split the sentence into words b) using loop, find the length of words c)and from that retrive the shortest word.
# 4. Logic : a) split the sentence into words. b) find the length of first word. c) if the length of first word is less than other word considered it as shortest word. d) then return the shortest word.
# 5. Code and Test:

sentence = "Python makes coding easy"
words = sentence.split()
smallest = words[0]

for word in words:
    if len(word) < len(smallest):
        smallest = word
print(smallest)
```

```python
# Q10. Find All Characters Appearing More Than Once

word = "programming"

# 1. Observe : Input - programming , Output - ['r', 'g', 'm']
# 2. Understanding : To find all characters from the above input.
# 3. Breakdown : a) write the Input b) use for loop and count the character from word greater than 1.
# 4. Logic : a) To get all characters appearing more than once using for loop, count() function.
# 5. Code and Test:

word = "programming"
result = []
for i in set(word):
    if word.count(i) > 1:
        result.append(i)
print(result)
```

```python
# Q15. Create a Dictionary of Numbers and Their Even/Odd Status

x = [1, 2, 3, 4, 5]

# 1. Observe : Input - [1,2, 3, 4, 5] , Output - {1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}
# 2. Understanding : To Create a Dictionary of Numbers and Their Even/Odd Status.
# 3. Breakdown : a) write the Input b) use for loop, create empty dictionary, divide the num by 2.
# 4. Logic : a) To Create a Dictionary of of Numbers and Their Squares using for loop,  by creating empty dictionary and divide the num by 2.
# 5. Code and Test :

x = [1, 2, 3, 4, 5]
d = {}
for i in x:
    if i % 2 == 0:
        d[i] = "Even"
    else:
        d[i]= "odd"
print(d)
```

```python
# Q17. Find All Words Appearing More Than Once

sent = "data science uses data and science tools"

# 1. Observe : Input - sent = "data science uses data and science tools" , Output - ['data', 'science']
# 2. Understanding : To find the Most Frequent Word in a Sentence.
# 3. Breakdown : 
#a) Take sent = "data science uses data and science tools" as Input 
#b) create an empty list and split the sentence in words using split() function. 
#c) using for loop, iterate through the value if word is greater than one append it to the empty list and print the new modified list. 

# 4. Logic : a) To find all words which is appearing more than once by using, split(), for loop method.
# 5. Code and Test :

sent = "data science uses data and science tools"
d = []
words = sent.split()
for word in words:
    if words.count(word)>1 and word not in d:
        d.append(word)
print(d)
```

```python
# Q19. Count Words Whose Length Is Greater Than 4.

sent = "Python makes problem solving interesting"

# 1. Observe : Input - sent = "Python makes problem solving interesting" , Output - 5.

# 2. Understanding : To Count Words Whose Length Is Greater Than 4.

# 3. Breakdown : 
#a) Take, sent = "Python makes problem solving interesting" as Input.
#b) Split the sent into words using split() function. 
#c) using for loop iterate through the value using len() function where words are greater than four, count those word.
#d) Print the counted word.

# 4. Logic : a) To Count Words Whose Length Is Greater Than 4 by using, split(), for loop, len() and count() function.

# 5. Code and Test :

sent = "Python makes problem solving interesting"
words = sent.split()
count = 0
for word in words:
    if len(word) > 4:
        count += 1

print(count)
```
