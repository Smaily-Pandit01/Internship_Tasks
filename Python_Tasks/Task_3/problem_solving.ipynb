```python
# Q1. Find the Frequency of Each Word in a Sentence

# Pseudo code:
# 1. Take a sentence as input from given question.
# 2. Take an empty dictionary to store the frequency of each word.
# 3. use for loop and split the sentence into words using the split() method.
# 4. count the frequency of each word and store it in the dictionary.
# 5. Print the dictionary containing the frequency of each word.

# Code Implementation :
sent = "python is easy and python is powerful"
d = {}

for word in sent.split():
    d[word] =sent.count(word)
print(d)

# Expected Output: {'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}

# Additional test case:
sent1 = "welcome to the world"

d1 = {}

for word in sent1.split():
    d1[word] =sent1.count(word)
print(d1)
```

```python
# Q6. Move All Zeros to the End of a List

# Pseudo code:
# 1. Take a list, list1 = [0, 3, 0, 1, 2, 0, 4] as input.
# 2. use for loop and iterate through the list and check if the element is equal to 0 then remove it from list and append it to the end of the list.
# 3. Print the modified list.

# Code implementation:
list1 = [0, 3, 0, 1, 2, 0, 4]

for i in list1:
    if i == 0:
        list1.remove(i)
        list1.append(i)
print(list1)

# Expected output:[3, 1, 2, 4, 0, 0, 0]

# Additional test case:
list2 = [0, 0, 1, 2, 3, 0]
for i in list2:
    if i == 0:
        list2.remove(i)
        list2.append(i)
print(list2)
```

```python
# Q7. Find the Second Highest Distinct Number in a List.

# Pseudo code:
# 1. Take a list, list1 = [10, 20, 40, 30, 40, 20] as input.
# 2. use set() method to remove the duplicate elements from the list.
# 3. use sorted() method to sort the list in ascending order and access the second last element of the list to get the second highest distinct number.
# 4. Print the second highest distinct number.

# code implementation:
list1 = [10, 20, 40, 30, 40, 20]
x = list(set(list1))
x.sort()

if len(x) >= 2:
    print(x[-2])
else:
    print(None)

# Expected output: 30

# Additional test case:
list2 = [5, 10, 15, 20, 25, 30]
y = list(set(list2))
y.sort()

if len(y) >= 2:
    print(y[-2])
else:
    print(None)
```

```python
# Q8. Remove Duplicates Without Changing the Original Order

# Pseudo code:
# 1. Take a list, a = [4, 2, 4, 5, 2, 7, 5] as input.
# 2. create an empty list to store the unique elements.
# 3. use for loop and iterate through the list and check if the element is not in the unique list then append it to the unique list.
# 4. Print the unique list.

# Code implementation:
a = [4, 2, 4, 5, 2, 7, 5]
b = []

for i in a:
    if i not in b:
        b.append(i)

print(b)

# Ecpected output: [4, 2, 5, 7]

# Additional test case:
a1 = [1, 2, 3, 2, 1, 4, 5]
b1 = []

for i in a1:
    if i not in b1:
        b1.append(i)

print(b1)
```

```python
# Q10. Find Two Numbers Whose Sum Equals a Target.

# Pseudo code:
# 1. Take a list, nums = [2, 7, 11, 15], target = 9 as input.
# 2. use for loop and iterate through the list if i and j are not equal and their sum is equal to target then print the numbers and break the loop.
# 3. else print empty list.

# code implementation:
nums = [2, 7, 11, 15]
target = 9

for i in nums:
    for j in nums:
        if i != j and i + j == target:
            print([i,j])
            break
    else:
        print([])
    break

# Expected output: [2, 7]

# Additional test case:
nums1 = [1, 2, 3, 4, 5]
target1 = 8

for i in nums1:
    for j in nums1:
        if i != j and i + j == target1:
            print([i,j])
            break
    else:
        print([])
    break
```

```python
# Q11. Find the Number Closest to Zero

# Pseudo code:
#1 . Take a list, a = [-8, -3, 2, 5, -2] as input.
#2. use min() method to find the minimum absolute value of the list use key=lambda x: abs(abs(x),-x) to get the closest positive number to zero.

# code implementation:
a = [-8, -3, 2, 5, -2]

print(min(a,key=lambda x:(abs(x),-x)))

# Expected output: 2

# Additional test case:
a1 = [-10, -5, 1, -1, 3]
print(min(a1,key=lambda x:(abs(x),-x)))

```

```python
# Q12. Find Common Elements Between Two Lists

# Pseudo code:
# 1 Take two lists, list1 = [1, 2, 3, 4, 5] and list2 = [3, 4, 5, 6, 7] as input.
# 2. create an empty list to store the common elements.
# 3. use for loop and iterate through the first list and check if the element is present in the second list then append it to the common elements list.
# 4. Print the common elements list.

# code implementation:
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

list3 = []

for i in list1:
    if i in list2:
        list3.append(i)

print(list3)

# Expected output: [3, 4, 5]

# Additional test case:
list4 = [10, 20, 30, 40]
list5 = [30, 40, 50, 60]

list6 = []

for i in list4:
    if i in list5:
        list6.append(i)

print(list6)
```

```python
# Q13. Rotate a List to the Right by K Positions

# Pseudo code:
# 1. Take a list, nums = [1, 2, 3, 4, 5] and k = 2 as input.
# 2. use loop and iterate through the range of k and use pop() method to remove the last element of the list and insert it at the beginning of the list using insert() method.
# 3. Print the modified list.

# Code implementation:
nums = [1, 2, 3, 4, 5]
k = 2

for i in range(k):
    x = nums.pop()
    nums.insert(0,x)

print(nums)

# Expected output: [4, 5, 1, 2, 3]

# Additional test case:
nums1 = [10, 20, 30, 40, 50]
k1 = 3

for i in range(k1):
    x = nums1.pop()
    nums1.insert(0,x)

print(nums1)
```

```python
# Q14. Find the Maximum Sum of Two Consecutive Elements

# Pseudo code:
# 1. Take a list, a = [4, 7, 2, 9, 5] as input.
# 2. add the ith and (i+1)th element and for loop in range of length of list -1 and store the sum in a new list.
# 3. use max() method to find the maximum sum from the new list.

# code implementation:
a = [4, 7, 2, 9, 5]

x = [a[i] + a[i+1] for i in range(len(a)-1)]

print(max(x))

# Expected output: 11

```

```python
# Q15. Find All Pairs of Numbers Whose Sum Equals a Target

# Pseudo code:
# 1. Take a list, nums = [2, 4, 3, 5, 7, 8, 1] and target = 9 as input.
# 2. create an empty list to store the pairs of numbers.
# 3. use for loop and iterate through the list and check if the sum of two numbers is equal to target then append the pair to the list.
# 4. Print the list of pairs.

nums = [2, 4, 3, 5, 7, 8, 1]
target = 9
pairs = []

for i in range(len(nums)):
    for j in range(i + 1,len(nums)):
        if nums[i] + nums[j] == target:
            pairs.append(sorted([nums[i],nums[j]]))

pairs.sort()
print(pairs)

# Expected output:[[1, 8], [2, 7], [4, 5]]
```
